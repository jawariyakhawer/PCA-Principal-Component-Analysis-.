```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("Mall_Customers.csv")
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
X = df.iloc[:, 2:5]
X.head()
```

```python
from sklearn.preprocessing import StandardScaler

sc = StandardScaler()

X_scaled = sc.fit_transform(X)
```

```python
from sklearn.decomposition import PCA
```

```python
pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_scaled)
```

```python
X_pca.shape
```

```python
print("Explained Variance Ratio:")
print(pca.explained_variance_ratio_)
```

```python
from sklearn.cluster import KMeans

# Determine the optimal number of clusters using elbow method or similar
# For demonstration, let's assume 5 clusters (a common choice for mall customer segmentation)
# In a real scenario, you would run an elbow method or silhouette score analysis first.
# For this example, let's create a KMeans model with a specific number of clusters.
# If you'd like to determine the optimal number of clusters, please let me know.

# Let's assume an optimal number of clusters, for instance, 5.
kmeans = KMeans(n_clusters=5, random_state=42, n_init=10) # n_init is set to suppress future warning
kmeans.fit(X_scaled)

# Create a DataFrame for the PCA results
principal_df = pd.DataFrame(data = X_pca, columns = ['PC1', 'PC2'])

# Add Cluster Labels
principal_df["Cluster"] = kmeans.labels_

plt.figure(figsize=(8,6))

plt.scatter(
    principal_df["PC1"],
    principal_df["PC2"],
    c=principal_df["Cluster"],
    cmap="viridis",
    s=70
)

plt.title("PCA with K-Means Clusters")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")

plt.colorbar(label="Cluster")

plt.show()
```

Conclusion

PCA successfully reduced the dimensionality of the dataset while retaining most of the important information. The transformed data can be visualized in two dimensions without significant information loss. PCA is especially useful as a preprocessing step before applying clustering or classification algorithms, as it improves efficiency and simplifies analysis.
